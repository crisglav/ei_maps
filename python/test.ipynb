```python
from neuromaps import datasets
import nibabel as nib
```

```python
# Fetch mni atlas
mni152 = datasets.fetch_atlas(atlas='MNI152', density='1mm')
print(sorted(mni152))
```

```python
# Load the brain mask of the mni152 template
brainmask = nib.load(mni152['2009cAsym_brainmask'])
print(brainmask.shape)
```

```python
# Fetch fsLR atlas
fslr = datasets.fetch_atlas(atlas='fsLR', density='32k')
print(sorted(fslr))
```

```python
fslr_left_infl = nib.load(fslr['inflated'].L)
vertices = fslr_left_infl.agg_data('NIFTI_INTENT_POINTSET')
print(vertices.shape[0])
```
