# Workflow for receptor parcellation and plotting as in Hansen et al 2022

First we load a sample receptor map, i.e. glutamate distribution as in Dubois 2015. This is a volumetric map in the MNI152 space. Then we parcellate the volumetric map to 100 regions of the Schaefer atlas with the neuromaps Parcellate module. Finally we plot the parcellated receptor distribution over the fsaverage brain.

```python
import numpy as np
from nilearn import datasets as nld
from nilearn import plotting as nlp

from neuromaps import datasets as nmd
from neuromaps.parcellate import Parcellater

from netneurotools import datasets as nnd
from netneurotools import plotting as nnp
```

```python
# Fetch Dubois from neuromaps
dubois = nmd.fetch_annotation(source='dubois2015', return_single=True)
```

```python
# Fetch Schaefer atlas from nilearn for parcellation
schaefer = nld.fetch_atlas_schaefer_2018(n_rois=100)
```

```python
# Parcellate Dubois to schaefer atlas
parcellater = Parcellater(schaefer['maps'], 'MNI152')
parcellated_dubois = parcellater.fit_transform(dubois, 'MNI152', True)
# Save parcellation to .csv
# np.savetxt('parcellated_dubois.csv', parcellated_dubois, delimiter=',')
```

```python
# Fetch Schaefer atlas using netneurotools for visualization
annot = nnd.fetch_schaefer2018('fsaverage')['100Parcels7Networks']
```

```python
# Plot parcellated PET map on fsavarege brain
brain = nnp.plot_fsaverage(data=parcellated_dubois.squeeze(),
                                lhannot = str(annot.lh),
                                rhannot = str(annot.rh),
                                colormap = 'plasma',
                                views=['lat','med'],
                                data_kws={'representation': "wireframe"})
brain.save_image('dubois.png')
```
